```python
# we are going implement a KNN 
# K= 3 for this case. 

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# we are going to use the data from titanic.csv
data = pd.read_csv('titanic.csv')
# we are going to seperate data into 30% test and 70% train
# Split the data int 30% test set and 70% train set. In the split function use a random state to 1 (what will this guarantee, state this in the report). 
```

Why did we set random_state = 1? 
- To ensure that the results are reproducible.
- To ensure that the results are consistent across different runs of the code. (This will remove the randomness and make the results consistent across different runs of the code.)

I will first remove the unnecessary features. (based on what sorts of people were more likely to survive)
we will remove 5 feqtures: passenger ID, Name,  Cabin, Ticket numeber, embarked Port of Embarkation. 

Here is reasons why I will remove these features:
1. Fare: They are co-related with Pclass, but as someone might have gotten discount or might have paid more for the same class, so it is a good idea to remove this feature. We don't want to have a feature that is co-related with another feature in our dataset which might cause multicollinearity.
2. Name: This is also just a unique arbitrary string assigned to each passenger at certain time in their life. This doesn't represent any information about the life of the passenger.
3. Cabin number: This is just a unique arbitrary string and number assigned to each passenger. This doesn't represent any information about the life of the passenger.
4. Ticket number: This is just a unique arbitrary string and number assigned to each passenger. This doesn't represent any information about the life of the passenger.
5. Embarked Port of Embarkation: This is where the passengers took the titanic from. They might have different socio-economic status, but this is not a good idea to look at where they came from as this only has 3 values and Pclass is a better representation of socio-economic status. This doesn't represent any information about the life of the passenger.


```python
# Now delete those 5 columns which are not useful for us
# PassengerId is not even our collection of data, and it is just index. Therefore, I will remove it in addition to the other columns.
data_modified = data.drop(columns=['PassengerId','Fare', 'Name', 'Ticket', 'Cabin', 'Embarked'])
```

```python
data_modified
```

Now what doe we have left?
We have the following features left:
1. Pclass: This is the ticket class of the passenger. This shows the socio-economic status of the passenger. 
2. Survived: This is the target variable. This shows whether the passenger survived or not.
3. Sex: This shows if they are male or female. 
4. Age: This shows the age of the passenger. 
5. SibSp: This shows the number of siblings/spouses aboard the Titanic.
6. Parch: This shows the number of parents/children aboard the Titanic.

Now, this is time to replace categorical values of a feature with numerical values

```python
# replace the categorical values with numerical values
# Let male = 0, and female = 1. 
data_modified['Sex'] = data_modified['Sex'].replace({'male': 0, 'female': 1})
# briefly check if the data is correct
print(data_modified.head())
```

Replace all missing values of numerical features with the mean of the values of the feature.

```python
# replace the missing values with the mean of the column
new_data_modified = data_modified.fillna(data_modified.mean())
print(new_data_modified.head())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, classification_report

# Now we are going to split the data into X and Y
X = new_data_modified.drop(columns=['Survived'])  
Y = new_data_modified['Survived']


# Now we are going to split the data into train and test
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=1)

# Now we are going to train the model
KNN = KNeighborsClassifier(n_neighbors=3)
KNN.fit(X_train, Y_train)

# make predictions
Y_pred = KNN.predict(X_test)

print("Confusion Matrix:")
print(confusion_matrix(Y_test, Y_pred))

print("\nClassification Report:")
print(classification_report(Y_test, Y_pred))

```
