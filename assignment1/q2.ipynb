```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# Load the dataset
data = pd.read_csv('housing.csv', delim_whitespace=True, header=None)
```

Why did we set random_state = 1? 
- To ensure that the results are reproducible. This fixes the seed of the random number generation, so that the results are consistent across different runs of the code.
- To ensure that the results are consistent across different runs of the code. (This will remove the randomness and make the results consistent across different runs of the code.)

# CASE A

```python
# Case A: Using all the data to predict the value of MEDV using Regression. 

X = data.iloc[:, :-1]  # Select all columns EXCEPT the last one (features)
Y = data.iloc[:, -1]   # Select ONLY the last column (target MEDV)

# Split the data
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=1)


from sklearn.linear_model import LinearRegression

# Scale the data
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train the model
lr = LinearRegression()
lr.fit(X_train_scaled, Y_train)

# Predictions
Y_pred_lr = lr.predict(X_test_scaled)

rmse = np.sqrt(mean_squared_error(Y_test, Y_pred_lr))
# Evaluate
print("Linear Regression:")
print(f"Root Mean Squared Error (RMSE): {rmse:}")
```

# CASE B

I need to choose what I will be dropping off as features (5 things)
I have dropped ZN, INDUS, CHAS, NOX, AGE as those are least relevant things to consider. 
Those are reasons why those are least relevant. 
1. ZN: Boston is urban area, and there is typically very little variation in the large lot residential zoning. Most of houses are already built / occupid, therefore, the zoning regulations doesn't provide much information into housing the prices. Therefore, this is low variance feature, and less informative to predict the MEDV.
2. INDUS: INDUS represents the extent of industrialization in the area. Although it is pretty important to consider the industrialization in the area, it is not the most important factor to consider when predicting the housing prices such as CRIM, LSTAT etc.. 
3. CHAS: This indicates whether a tract borders the Charles River or not. And, this is only represented as binary value. Also,only a small fraction of boston are bordering the river. Therefore, this is not a strong indicator to predict the housing prices.
4. NOX: This indicates the pollution level of the environment. And, theoritically it is important to consider the pollution level of the environment when predicting the MEDV. However, as it might be non-liner and complext reationship, it is not the most important factor to consider when predicting the housing prices. (such as people usually don't really know the pollution level of the environment when they are buying the house, but also at the same time some peopel do. This will be a complex relationship to predict the housing prices.)
5. AGE: This is a bit ambiguous as older homes mibhg be more expensive due to the historical value, but also at the same time, it might be cheaper due to outdated facilities. Therefore, this is not the most important factor to consider when predicting the housing prices.

```python
# Case B: Remove 5 of the features that you consider to be least relevant (explain your selection).
import random
x = data.iloc[:, :-1]  # Select all columns EXCEPT the last one (features)
y = data.iloc[:, -1]   # Select ONLY the last column (target MEDV)


# now drop 5 feautres that are least relevant 
# ill drop ZN, INDUS, CHAS, NOX, AGE, but those dont have headers so i will drop them by index
x_dropped = x.drop(x.columns[[1, 2, 3, 4, 6]], axis=1)

x_train, x_test, y_train, y_test = train_test_split(x_dropped, y, test_size=0.3, random_state=1)

from sklearn.linear_model import LinearRegression
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)

lr = LinearRegression()
lr.fit(x_train_scaled, y_train)

y_pred_lr = lr.predict(x_test_scaled)

rmse = np.sqrt(mean_squared_error(y_test, y_pred_lr))
# Evaluate
print("Linear Regression:")
print(f"Root Mean Squared Error (RMSE): {rmse:}")
```

As those two case A and case B has simliar RMSE, this conveys that removing 5 features didn't affect the linear regression model. Therefore, it is better to remove those 5 features as it will make the model simpler and easier to interpret. Those are low informative features.
